```python
# QUESTION 3

def calculate_discount(price, discount_percentage):
  discount_amount = price * (discount_percentage / 100)
  discounted_price = price - discount_amount
  return discounted_price
discount_percentage = 10
calculate_discount(30000, discount_percentage)
```

```python
# QUESTION 6

def factorial(n):
  if n == 0 or n == 1:
    return 1
  else:
    return n * factorial(n-1)

print(factorial(5))
print(factorial(0))
print(factorial(1))
```

```python
# QUESTION 7

def apply_function(func, numbers):
  results = []
  for num in numbers:
    results.append(func(num))
  return results


def square(x):
  return x * x

my_numbers = [1, 2, 3, 4, 5]
squared_numbers = apply_function(square, my_numbers)
print(squared_numbers)
```

```python
# QUESTION 9

def apply_operation(func, numbers):
  results = []
  for num in numbers:
    results.append(func(num))
  return results


def double(x):
  return x * 2

my_list = [1, 2, 3, 4, 5]
doubled_list = apply_operation(double, my_list)
print(doubled_list)
```

```python
# QUESTION 10

import time

def measure_time(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        duration = end_time - start_time
        print(f"Function '{func.__name__}' executed in {duration:.4f} seconds.")
        return result
    return wrapper



@measure_time
def greet(name):
    time.sleep(0.5) # Simulate some work
    return f"Hello, {name}!"

print(f"Result of greet: {greet('NAMEERA')}")
```
